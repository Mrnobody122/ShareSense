```python
# Question 2
def find_sum(numbers):
  sum = 0
  for num in numbers:
    sum += num
  return sum

def avg(num):
  total = find_sum(num)
  avg = total/len(num)
  return avg

def maximum(num):
  max_value = max(num)
  return max_value

def minimum(num):
  min_value = min(num)
  return min_value

def main():
  print('Enter your choice: 1 for sum\n2 for average\n3 for maximum\n4 for minimum')
  try:
    choice = int(input('Enter your choice: '))
    nums = input('Enter numbers seperate by spaces: ')
    if nums.strip() == "":
      raise Exception('Invalid input')
    number = []
    split_nums = nums.split()
    if len(split_nums) == 0:
      raise Exception('Invalid input')

    for num in split_nums:
      number.append(int(num))

    if choice == 1:
      print("Sum:", find_sum(number))

    elif choice == 2:
      print("Average:", avg(number))

    elif choice == 3:
      print("Maximum:", maximum(number))

    elif choice == 4:
      print("Minimum:", minimum(number))

    else:
      raise ValueError("Invalid choice")

  except ValueError:
    print('Invalid input')
main()
```

```python
# Question 4
def get_sublist(lst, start, end):
    """
    Returns a sublist from the given list starting from index 'start'
    and ending at index 'end' (inclusive).

    Parameters:
    lst (list): The original list
    start (int): Starting index
    end (int): Ending index

    Returns:
    list: The extracted sublist
    """

    sublist = []

    for i in range(start, end + 1):
        sublist.append(lst[i])

    return sublist

numbers = [1, 2, 3, 4, 5, 6]
result = get_sublist(numbers, 2, 4)

print(result)
```

```python
def flatten(lst):
  flat_list=[]

  for sublist in lst:
    for item in sublist:
      flat_list.append(item)
  return flat_list

nested = [[1, 2], [3, 4], [5]]
result = flatten(nested)

print(result)
```

```python
def access_nested_element(lst, indices):
    element = lst

    for index in indices:
        element = element[index]

    return element

lst = [[1,2,3], [4,5,6], [7,8,9]]
indices = [1,2]

result = access_nested_element(lst, indices)
print(result)
```

```python
def remove_element(lst, elem):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.append(remove_element(item, elem))
        else:
            if item != elem:
                result.append(item)

    return result

lst = [[1, 2], [3, 2], [4, 5]]
elem = 2

result = remove_element(lst, elem)
print(result)
```

```python
def sum_nested(lst):
  total_sum = 0
  for element in lst:
    if isinstance(element, list):
      total_sum += sum_nested(element)
    else:
      total_sum += element
  return total_sum

# Example usage:
nested_list = [[1, 2], [3, [4, 5]], 6]
result = sum_nested(nested_list)
print(f"The sum of all elements in {nested_list} is: {result}")

nested_list_2 = [1, [2, [3, 4], 5], 6]
result_2 = sum_nested(nested_list_2)
print(f"The sum of all elements in {nested_list_2} is: {result_2}")
```

```python
def find_max(lst):
    maximum = float('-inf')

    for item in lst:
        if isinstance(item, list):
            value = find_max(item)
            maximum = max(maximum, value)
        else:
            maximum = max(maximum, item)

    return maximum

nested_list = [[1, 2], [3, [4, 5]], 6]

result = find_max(nested_list)
print(result)
```

```python
def count_occurrences(lst, elem):
    count = 0

    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        else:
            if item == elem:
                count += 1

    return count

lst = [[1, 2], [2, 3], [2, 4]]
elem = 2

result = count_occurrences(lst, elem)
print(result)
```

```python
def average_nested(lst):
    total = 0
    count = 0

    for item in lst:
        if isinstance(item, list):
            sub_total, sub_count = average_nested(item)
            total += sub_total
            count += sub_count
        else:
            total += item
            count += 1

    return total, count

def get_average(lst):
    total, count = average_nested(lst)
    return total / count

nested_list = [[1, 2], [3, 4], [5, 6]]

result = get_average(nested_list)
print(result)
```

```python
import numpy as np

# Matrix operations
A = np.array([[3, 4],
              [7, 8]])

B = np.array([[5, 3],
              [2, 1]])

# 1. Prove A * A^-1 = I
A_inv = np.linalg.inv(A)
print("A inverse:")
print(A_inv)

print("\nA * A_inv:")
print(np.dot(A, A_inv))

# 2. Prove AB != BA
AB = np.dot(A, B)
BA = np.dot(B, A)

print("\nAB =")
print(AB)

print("\nBA =")
print(BA)

# 3. Prove (AB)^T = B^T A^T
print("\n(AB)^T =")
print(AB.T)

print("\nB^T A^T =")
print(np.dot(B.T, A.T))

# Linear equation solving
A_eq = np.array([[2, -3, 1],
                 [1, -1, 2],
                 [3,  1, -1]])

B_eq = np.array([[-1],
                 [-3],
                 [9]])

# Solve using inverse method
X = np.dot(np.linalg.inv(A_eq), B_eq)

print("\nSolution of the system:")
print(X)
```

```python
# Create two 1000x1000 matrices using Python lists
size = 1000

A = [[1 for _ in range(size)] for _ in range(size)]
B = [[1 for _ in range(size)] for _ in range(size)]

# Result matrix
C = [[0 for _ in range(size)] for _ in range(size)]

start = time.time()

for i in range(size):
    for j in range(size):
        for k in range(size):
            C[i][j] += A[i][k] * B[k][j]

end = time.time()

print("Time taken using Python Lists:", end - start)
```
